**BookVoice-AI — Lied/Gesang Generator Test**

Testet ob Bark türkisch/deutsch singen kann

```
#@title ⚙️ Schritt 1: Bark installieren (~2GB)
import subprocess, sys
print('Bark installieren...')
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'bark'], capture_output=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'scipy'], capture_output=True)
print('OK!')

from bark import generate_audio, SAMPLE_RATE, preload_models
import os
os.environ['SUNO_USE_SMALL_MODELS'] = '0'
os.environ['SUNO_OFFLOAD_CPU'] = '0'
print('Modell laden...')
preload_models()
print('Fertig!')
```

```
#@title 🎵 Schritt 2: Türkisch singen
text_prompt = '♪ Bismillah, Rahman ve Rahim, Alemlerin Rabbi Allah ♪' #@param {type:'string'}
stimme = 'v2/tr_speaker_0' #@param ['v2/tr_speaker_0', 'v2/tr_speaker_1', 'v2/tr_speaker_2', 'v2/tr_speaker_3', 'v2/tr_speaker_4']

import scipy.io.wavfile as wavfile
import numpy as np
from IPython.display import Audio, display
from bark import generate_audio, SAMPLE_RATE

print(f'Generiere: {text_prompt}')
audio = generate_audio(text_prompt, history_prompt=stimme)
wavfile.write('/content/bark_test_tr.wav', SAMPLE_RATE, audio)
print('Fertig! Anhören:')
display(Audio('/content/bark_test_tr.wav'))
```

```
#@title 🎵 Schritt 3: Deutsch singen
text_prompt_de = '♪ Guten Morgen, schöner Tag, die Sonne scheint so hell ♪' #@param {type:'string'}
stimme_de = 'v2/de_speaker_0' #@param ['v2/de_speaker_0', 'v2/de_speaker_1', 'v2/de_speaker_2', 'v2/de_speaker_3']

from bark import generate_audio, SAMPLE_RATE
import scipy.io.wavfile as wavfile
from IPython.display import Audio, display

print(f'Generiere: {text_prompt_de}')
audio = generate_audio(text_prompt_de, history_prompt=stimme_de)
wavfile.write('/content/bark_test_de.wav', SAMPLE_RATE, audio)
print('Fertig! Anhören:')
display(Audio('/content/bark_test_de.wav'))
```

```
#@title 🎤 Schritt 4: Verschiedene Stile testen
#@markdown Teste verschiedene Bark Spezial-Tags

from bark import generate_audio, SAMPLE_RATE
import scipy.io.wavfile as wavfile
from IPython.display import Audio, display

tests = {
    'Normal singen TR': '♪ Ellerim sende, kalbim sende ♪',
    'Fluestern TR': '[clears throat] Bismillah... [whispers] Rahman Rahim',
    'Lachen + singen': '[laughs] ♪ Bu gece cok guzel ♪',
    'Dramatisch TR': '[gasps] Allahu Akbar! [sighs heavily]',
}

for name, text in tests.items():
    print(f'\n--- {name} ---')
    audio = generate_audio(text, history_prompt='v2/tr_speaker_0')
    path = f'/content/bark_{name.replace(" ","_")}.wav'
    wavfile.write(path, SAMPLE_RATE, audio)
    display(Audio(path))
```
