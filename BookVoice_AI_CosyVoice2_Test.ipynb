```
#@title 🎵 Schritt 3: Modell laden und testen
import sys, os
sys.path.insert(0, '/content/CosyVoice')
sys.path.insert(0, '/content/CosyVoice/third_party/Matcha-TTS')
os.environ['COQUI_TOS_AGREED'] = '1'

from cosyvoice.cli.cosyvoice import CosyVoice2
from cosyvoice.utils.file_utils import load_wav
import torchaudio
from IPython.display import Audio, display

print('Modell laden...')
model = CosyVoice2('pretrained_models/CosyVoice2-0.5B', 
    load_jit=False, load_trt=False)
print('OK!')
```

```
#@title 🎤 Schritt 4: Stimme hochladen und testen
from google.colab import files
import torchaudio
from IPython.display import Audio, display

print('Stimme hochladen (WAV/MP3):')
uploaded = files.upload()
voice_file = list(uploaded.keys())[0]
print(f'Stimme: {voice_file}')

# Test Text
test_text = 'Bismillahirrahmanirrahim. Rahman ve Rahim olan Allahın adıyla.'

prompt_speech_16k = load_wav(voice_file, 16000)

print('Generiere...')
for i, result in enumerate(model.inference_zero_shot(
    test_text, 
    'Bu bir test kaydidir.',
    prompt_speech_16k, 
    stream=False
)):
    torchaudio.save(f'/content/cosy_test_{i}.wav', result['tts_speech'], model.sample_rate)
    print(f'Fertig! Anhören:')
    display(Audio(f'/content/cosy_test_{i}.wav'))
```
